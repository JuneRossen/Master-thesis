# General aim
This notebook was developed as part of the Master’s thesis of June Rossen and is the last of five notebooks used to prepare data for a wind power placement multi-objective optimization.

This fifth notebook regionalizes a nation-level grid containing information for the optimization into the five electricity price regions of Norway. It then saves the results into new region grids that can be used for the optimization. 

```python
import pandas as pd
import geopandas as gpd
import numpy as np
```

```python
# Load vector and csv data
NO_grid = gpd.read_file("data/SSB_techlegalfiltered_opticriteria_normalized_merged.gpkg")

NO2 = gpd.read_file("data/ElSpot/Polygon.shp")
NO5 = gpd.read_file("data/ElSpot/Polygon1.shp")
NO1 = gpd.read_file("data/ElSpot/Polygon2.shp")
NO3 = gpd.read_file("data/ElSpot/Polygon3.shp")
NO4 = gpd.read_file("data/ElSpot/Polygon4.shp")
```

```python
print(NO_grid.crs, NO2.crs, NO2.crs, NO2.crs, NO2.crs, NO2.crs)
```

```python
NO_grid = NO_grid.to_crs(NO2.crs)
```

```python
regions = [NO1, NO2, NO3, NO4, NO5]
regions_names = ["NO1", "NO2", "NO3", "NO4", "NO5"]
```

```python
# Split the national grid into the regions and save into files
array_filtered_grid = []
for i, r in enumerate(regions):
    region = r.geometry.iloc[0]
    grid_filtered = NO_grid[NO_grid.geometry.centroid.within(region)]
    array_filtered_grid.append(grid_filtered)

    file_name = "data/" + regions_names[i] + "_normalized_merged.gpkg"
    grid_filtered.to_file(file_name, driver="GPKG")
```
